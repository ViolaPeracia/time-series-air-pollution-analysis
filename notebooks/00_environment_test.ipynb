```python
import pandas as pd
import numpy as np
import pyarrow as pa
import matplotlib
import matplotlib.pyplot as plt
import seaborn as sns
import scipy
import statsmodels
import sklearn

print("All core libraries imported successfully.")
```

```python
libraries = {
    "pandas":      pd.__version__,
    "numpy":       np.__version__,
    "pyarrow":     pa.__version__,
    "matplotlib":  matplotlib.__version__,
    "seaborn":     sns.__version__,
    "scipy":       scipy.__version__,
    "statsmodels": statsmodels.__version__,
    "scikit-learn": sklearn.__version__,
}

version_df = pd.DataFrame(
    list(libraries.items()),
    columns=["Library", "Version"],
)
print(version_df.to_string(index=False))
```

```python
# Smoke test: create a small DataFrame and save/load as Parquet
import tempfile, os

df_test = pd.DataFrame({
    "timestamp": pd.date_range("2024-01-01", periods=5, freq="h"),
    "pm25":      [35.2, 42.1, 38.7, 55.0, 47.3],
})

tmp_path = os.path.join(tempfile.gettempdir(), "env_test.parquet")
df_test.to_parquet(tmp_path, engine="pyarrow", compression="snappy")
df_reload = pd.read_parquet(tmp_path, engine="pyarrow")

assert df_test.shape == df_reload.shape, "Parquet round-trip shape mismatch!"
assert (df_test["pm25"] == df_reload["pm25"]).all(), "Parquet round-trip data mismatch!"

os.remove(tmp_path)
print("Parquet round-trip test PASSED.")
```

```python
# Smoke test: verify matplotlib can render a figure
fig, ax = plt.subplots(figsize=(4, 2))
ax.plot([1, 2, 3], [1, 4, 9], marker="o")
ax.set_title("Matplotlib render test")
plt.close(fig)
print("Matplotlib render test PASSED.")
```

```python
# Smoke test: scikit-learn pipeline
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import RobustScaler

pipe = Pipeline([("scaler", RobustScaler())])
X = np.array([[1.0], [2.0], [3.0], [100.0]])
X_scaled = pipe.fit_transform(X)

assert X_scaled.shape == X.shape, "Pipeline output shape mismatch!"
print("Scikit-learn Pipeline test PASSED.")
```
